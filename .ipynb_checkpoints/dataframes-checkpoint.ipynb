```python
import numpy as np
import pandas as pd
```

```python
data_frame = pd.read_csv('CAISO.csv', index_col = 0)
data_frame
```

```python
power_data = data_frame.T
power_data
```

```python
import matplotlib
power_data.plot(kind='line', title='CAISO hourly data', xlabel='Time', ylabel='MW')
```

```python
renewables = power_data['Renewables']
print('Renewables')
print('Maximum:', renewables.max())
print('Maximum:', renewables.min())
print('Average:', renewables.mean())
```

```python
renewables_GW = renewables / 1000
print('Maximum:', renewables_GW.max(), 'GW')
print('Maximum:', renewables_GW.min(), 'GW')
print('Average:', renewables_GW.mean(), 'GW')
```
